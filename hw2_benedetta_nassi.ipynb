```python
from tensorflow.keras.datasets import cifar10
from sklearn.preprocessing import StandardScaler

(X_train, y_train), (X_test, y_test) = cifar10.load_data()


y_train = y_train.flatten()
y_test = y_test.flatten()


X_train, y_train = shuffle(X_train, y_train, random_state=42)
X_test, y_test = shuffle(X_test, y_test, random_state=42)

X_train = X_train.reshape((X_train.shape[0], -1))
X_test = X_test.reshape((X_test.shape[0], -1))


scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
```

```python

from sklearn.model_selection import train_test_split

X_train_full = X_train
y_train_full = y_train

# Sottocampiono il 10% di CIFAR-10
n_percent = 0.1
X_train, _, y_train, _ = train_test_split(
    X_train_full, y_train_full,
    train_size= n_percent,
    random_state=42
)

print("Shape of X:", X_train.shape)
print("Shape of y:", y_train.shape)
```

```python

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report

def metriche (model, X_test, y_test):
    predictions = model.predict(X_test)
    accuracy = accuracy_score(y_test, predictions)
    precision = precision_score(y_test, predictions, average='weighted')
    recall = recall_score(y_test, predictions, average='weighted')
    f1 = f1_score(y_test, predictions, average='weighted')
    return (accuracy, precision, recall, f1)
```

```python
# Regressione Logistica
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV


param_grid_log = {
    'C': [0.1, 1, 10]
}
grid_log = GridSearchCV(LogisticRegression(max_iter=500), param_grid_log, cv=4, scoring='accuracy')
grid_log.fit(X_train, y_train)

print("Miglior modello Logistic Regression:", grid_log.best_params_)
log_accuracy = grid_log.best_score_ 
print(f"Accuracy (CV): {log_accuracy}")

#Metriche e prestazioni
log_model = grid_log.best_estimator_
log_test_accuracy, log_test_precision, log_test_recall, log_test_f1 = metriche(log_model, X_test, y_test)
print(f"Accuracy: {log_test_accuracy}")
print(f"Precision: {log_test_precision}")
print(f"Recall: {log_test_recall}")
print(f"F1: {log_test_f1}\n")

```

```python
#SVM
from sklearn.svm import SVC

param_grid_svm = {
    'C': [0.1, 1, 10]
}
grid_svm = GridSearchCV(SVC(), param_grid_svm, cv=4, scoring='accuracy')
grid_svm.fit(X_train, y_train)


print("Miglior modello SVM:", grid_svm.best_params_)
svm_accuracy = grid_svm.best_score_
print(f"Accuracy (CV): {svm_accuracy}")

svm_model = grid_svm.best_estimator_

# Metriche e prestazioni
svm_accuracy, svm_precision, svm_recall, svm_f1 = metriche (svm_model, X_test, y_test)
print(f"Accuracy: {svm_accuracy}")
print(f"Precision: {svm_precision}")
print(f"Recall: {svm_recall}")
print(f"F1: {svm_f1}")
```

```python
#knn
from sklearn.neighbors import KNeighborsClassifier

param_grid_knn = {
    'n_neighbors': [3, 5, 7, 9],
    'weights': ['uniform', 'distance']
}
grid_knn = GridSearchCV(KNeighborsClassifier(), param_grid_knn, cv=4, scoring='accuracy')
grid_knn.fit(X_train, y_train)

print("Miglior modello KNN:", grid_knn.best_params_)
knn_accuracy = grid_knn.best_score_
print(f"Accuracy (CV): {knn_accuracy}")

knn = grid_knn.best_estimator_

# Metriche e prestazioni
knn_accuracy, knn_precision, knn_recall, knn_f1 = metriche (knn, X_test, y_test)
print(f"Accuracy: {knn_accuracy}")
print(f"Precision: {knn_precision}")
print(f"Recall: {knn_recall}")
print(f"F1: {knn_f1}")
```

```python
#decision trees
from sklearn.tree import DecisionTreeClassifier


from sklearn.tree import DecisionTreeClassifier
param_grid_tree = {
    'max_depth': [3, 5, 7, None],
    'min_samples_split': [2, 5, 10]
}
grid_tree = GridSearchCV(DecisionTreeClassifier(), param_grid_tree, cv=4, scoring='accuracy')
grid_tree.fit(X_train, y_train)


print("Miglior modello Decision Tree:", grid_tree.best_params_)
tree_accuracy = grid_tree.best_score_
print(f"Accuracy (CV): {tree_accuracy}")

tree = grid_tree.best_estimator_

# Metriche e prestazioni
tree_accuracy, tree_precision, tree_recall, tree_f1 = metriche (tree, X_test, y_test)
print(f"Accuracy: {tree_accuracy}")
print(f"Precision: {tree_precision}")
print(f"Recall: {tree_recall}")
print(f"F1: {tree_f1}")
```
